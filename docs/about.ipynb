# Franck DEBA

Je suis Franck Deba, ingénieur Data Scientist diplômé de l’ENSAI et
actuellement étudiant au M2MO (ex-DEA Laure-Elie). Je me spécialise en
finance de marché après un parcours en gestion des risques bancaires.

Ce site web, conçu avec Quarto, s’adresse principalement aux étudiants
intéressés par la finance de marché et ses métiers. L’accent est mis sur
la pédagogie pour appréhender les défis académiques que j’ai rencontrés
et que beaucoup vivent durant leur formation.

Il ne prétend en aucun cas se substituer à un cours - la rigueur et
l’expertise des enseignants restent essentielles. Son objectif est
plutôt de compléter leur travail en partageant mes expériences
personnelles et les projets que j’ai réalisés.

Au-delà des étudiants, ce site constitue également une ressource pour
les professionnels souhaitant revisiter certains concepts clés en
gestion des risques ou en statistiques. Il propose des exemples concrets
et des pistes de réflexion adaptées aux problématiques actuelles.
