# Question 3: Mise à jour de la base de données d'articles
Ici nous mettons à jours la base de données avec le texte des articles du flux RSS. Nous avons fait le choix de conserver la même structure des article que celle de la base de données existante.

- 'author' : auteur de l'article,
- 'category': la catégorie de l'article (france, europe, économie, etc.)
- 'content' : Le contenue de l'article,
- 'date': date de publication de l'article,
- 'image_link': lien vers l'image d'illustration utilisé dans l'article,
- 'title': Titre de l'article

L'identification de ces nouveaux articles c'est faite par comparaison de leur lien url avec ceux des articles de la base de données.



````python
import json
import ssl
import feedparser as fp
import newspaper as np
from pprint import pprint
import pandas as pd

# (a) Prévenir les problèmes SSL
ssl._create_default_https_context = ssl._create_unverified_context

# Fonction pour mettre à jour la base de données avec les articles du flux RSS
def mettre_a_jour_base(flux_rss_dict, base_donnees):
    """
    Met à jour une base de données avec les articles récupérés depuis les flux RSS.

    :param flux_rss_dict: Dictionnaire avec les catégories comme clés et URLs RSS comme valeurs
    :param base_donnees: Dictionnaire contenant les articles existants (clé = lien)
    :return: (base mise à jour, articles ajoutés)
    """
    # Convertir les clés existantes en un ensemble pour éviter les doublons
    articles_existants = list(base_donnees.keys())

    # Nouveau dictionnaire pour les articles à ajouter
    nouveaux_articles = {}

    # Parcourir les flux RSS
    for category, url_flux in flux_rss_dict.items():
        print(f"Lecture du flux RSS '{category}' : {url_flux}")
        data = fp.parse(url_flux)

        # Parcourir chaque article du flux
        for item in data.entries:
            lien = item.get("link", "")
            if lien and lien not in articles_existants:
                try:
                    # Extraire les informations détaillées avec newspaper
                    article_obj = np.Article(lien)
                    article_obj.download()
                    article_obj.parse()

                    # Ajouter l'article au dictionnaire avec la clé 'category'
                    nouveaux_articles[lien] = {
                        "title": article_obj.title,
                        "author": article_obj.authors,
                        "date": article_obj.publish_date.isoformat(),
                        "content": article_obj.text,
                        "image_link": article_obj.top_image,
                        "category": category  # Ajouter la catégorie correspondant au flux
                    }
                    print(f"Article ajouté depuis '{category}': {lien}")
                except Exception as e:
                    print(f"Erreur pour l'article {lien} : {e}")

    # Mettre à jour la base de données existante
    base_donnees.update(nouveaux_articles)
    print(f"{len(nouveaux_articles)} articles ajoutés à la base.")
    return base_donnees, nouveaux_articles

a,b = mettre_a_jour_base(tv_info_source_data, france_tv_data)






````

# Question 7: Technique de détection des entités nommées

Les étiquettes IOB (Inside-Outside-Beginning) permettent de marquer chaque mot d’un texte en fonction de sa relation avec une entité nommée. Par exemple, dans la phrase "Emmanuel Macron est à Mayotte", "Emmanuel" et "Macron" pourraient être étiquetés respectivement comme B-PER (Beginning of a Person) et I-PER (Inside a Person), tandis que "Mayotte" serait étiqueté B-LOC (Beginning of a Location).

Les modèles de machine learning, comme les CRF (Conditional Random Fields) ou les réseaux LSTM , sont utilisés pour cette tâche en apprenant à prédire ces étiquettes à partir d’un ensemble de données étiquetées. Ces modèles utilisent des embeddings de mots pour mieux capturer le contexte et les relations entre les mots afin d'identifier avec précision les entités nommées dans un texte.

Une technique de détection des entités nommées consiste à attribuer à chaque mot de la phrase une étiquette de façon aléatoire et de mettre en place un modèle de RNN  qui va servir à prédire les différentes étiquettes pour les mots. Les modèles plus récents, comme les transformers (par exemple, BERT et GPT), ont une meilleure comprenhension du contexte global des phrases.

# Question 8: Détection des entités nommées

Ici nous commencer par charger le modèle fr_core_news_md pour le traitement du texte en français, et la pipeline senter est activé pour la segmentation des phrases.

La fonction extraire_entites_et_verbes définie prend en entrée la base de données des articles et retourne un dictionnaire d'articles enrichis avec les entités nommées et les verbes extraits du contenu des articles.

Pour chaque article dans base_articles :
Le contenu textuel de l'article est récupéré.
Si le texte est présent, il est traité avec le modèle nlp.


Les entités nommées sont extraites du texte et stockées dans une liste de dictionnaires, chaque dictionnaire contenant le texte de l'entité, son type, et ses positions de début et de fin dans le texte.

Les phrases sont extraites du contenu de l'article
et pour chaque phrase, les verbes sont identifiés et extraits sous forme de lemmes (formes de base des verbes).
Si une phrase contient des verbes ou des entités nommées, elle est ajoutée à la liste des phrases avec les informations sur les verbes et les entités.

- Construction du dictionnaire enrichi :

Pour chaque article, un dictionnaire est créé avec les informations suivantes : titre, auteur, date, contenu, lien de l'image, catégorie, entités nommées, et **phrases contenant des verbes et des entités**.

----
Nous avons constaté que la désactivation de certains traitements dans la chaîne de traitement de SpaCy, bien qu’ils ne soient pas directement utiles pour les tâches à accomplir, influence les sorties des entités au format IOB. Cela révèle que les différents traitements de la méthode nlp ne sont pas totalement indépendants, ce qui a motivé notre choix de conserver l’ensemble des composants activés.

Par exemple, en désactivant le module d’embeddings, l’entité "Emmanuel Macron" n’est plus reconnue comme une entité unique. À la place, deux entités distinctes, "Emmanuel" et "Macron", sont restituées. Ce résultat met en évidence une perte d’information, car le contexte global du texte n’est plus suffisamment pris en compte dans les sorties. Cette observation souligne l’importance d’une approche contextuelle pour préserver la cohérence et la précision des entités identifiées.

````python
import json
import spacy


# Charger le modèle sans rien désactiver
nlp = spacy.load("fr_core_news_md")

# Activer le senter pour la segmentation
nlp.enable_pipe("senter")

def extraire_entites_et_verbes(base_articles):

  """
  Fonction pour extraire les entités nommées et les verbes d'un ensemble d'articles
  Entrée : base_articles - dictionnaire contenant les articles avec leur contenu textuel
  Sortie : dictionnaire enrichi avec les entités nommées et les verbes extraits

  """
  articles_enrichis = {}
  for lien, article in base_articles.items():
      texte = article.get("content", "")
      if texte:
          doc = nlp(texte)

          # Extraire les entités nommées
          entites = [
                {"entite": ent.text, "type": ent.label_, "start": ent.start_char, "end": ent.end_char}
                for ent in doc.ents
            ]

          # Extraire les phrases
          phrases = []
          for sent in doc.sents:
              # Extraire les verbes
              verbes = [
                    {"verbe": token.lemma_, "start": token.idx, "end": token.idx + len(token)}
                    for token in sent if token.pos_ == "VERB"
                ]
              # Ajout de la phrase si verbes ou entités
              if verbes or sent.ents:
                  phrases.append({
                        "texte": sent.text.strip(),
                        "start": sent.start_char,
                        "end": sent.end_char,
                        "verbes": verbes,
                        "entites": [
                            {
                                "entite": ent.text,
                                "type": ent.label_,
                                "start": ent.start_char,
                                "end": ent.end_char
                            }
                            for ent in sent.ents
                        ]
                    })

          articles_enrichis[lien] = {
                "title": article.get("title", ""),
                "author": article.get("author", ""),
                "date": article.get("date", ""),
                "content": texte,
                "image_link": article.get("image_link", ""),
                "category": article.get("category", ""),
                "entities": entites,
                "phrases": phrases
            }
  return articles_enrichis





#test = first_three_elements = dict(list(france_tv_data.items())[:3])
#article_enrichis_test = extraire_entites_et_verbes(test)
articles_enrichis = extraire_entites_et_verbes(france_tv_data)
````

# Question 9: les 20 entités les plus fréquentes dans la collection de documents (par types d'entités)

Les fonctions ci-dessous ont pour objectif d'analyser les entités nommées dans un ensemble d'articles et de déterminer les 20 entités les plus fréquentes pour chaque type d'entité (personnes, lieux, organisations).

````python
import json
from collections import Counter

# Fonction pour déterminer les 20 entités les plus fréquentes par type
def top_entites_par_type(fichier):
    """
    Analyse les entités nommées et détermine les 20 plus fréquentes par type.
    :param fichier_json: Chemin vers le fichier JSON contenant les entités enrichies.
    :return: Trois listes avec les entités les plus fréquentes pour PER, LOC, et ORG.
    """
    # Compteurs pour chaque type d'entité
    compteur_per = Counter()
    compteur_loc = Counter()
    compteur_org = Counter()

    # Charger les données enrichies

    base_enrichie = fichier

    # Parcourir chaque article pour collecter les entités
    for article in base_enrichie.values():
        for entite in article.get("entities", []):
            if entite["type"] == "PER":
                compteur_per[entite["entite"]] += 1
            elif entite["type"] == "LOC":
                compteur_loc[entite["entite"]] += 1
            elif entite["type"] == "ORG":
                compteur_org[entite["entite"]] += 1

    # Obtenir les 20 entités les plus fréquentes pour chaque type
    top_per = compteur_per.most_common(20)
    top_loc = compteur_loc.most_common(20)
    top_org = compteur_org.most_common(20)

    return top_per, top_loc, top_org


def afficher_top_entites(top_per, top_loc, top_org):
    """
    Affiche proprement les entités les plus fréquentes pour chaque type (PER, LOC, ORG).
    :param top_per: Liste des entités 'PER' (Personnes) les plus fréquentes.
    :param top_loc: Liste des entités 'LOC' (Lieux) les plus fréquentes.
    :param top_org: Liste des entités 'ORG' (Organisations) les plus fréquentes.
    """
    print("\n" + "=" * 50)
    print(f"{'Top 20 des entités de type PER (Personnes)':^50}")
    print("=" * 50)
    for i, (entite, freq) in enumerate(top_per, start=1):
        print(f"{i:>2}. {entite:<30} - {freq} occurrence(s)")

    print("\n" + "=" * 50)
    print(f"{'Top 20 des entités de type LOC (Lieux)':^50}")
    print("=" * 50)
    for i, (entite, freq) in enumerate(top_loc, start=1):
        print(f"{i:>2}. {entite:<30} - {freq} occurrence(s)")

    print("\n" + "=" * 50)
    print(f"{'Top 20 des entités de type ORG (Organisations)':^50}")
    print("=" * 50)
    for i, (entite, freq) in enumerate(top_org, start=1):
        print(f"{i:>2}. {entite:<30} - {freq} occurrence(s)")






````

# Question 10: Co-occurence d'entités au sein du même document

Cette fonction permet d'analyser les relations entre différentes entités dans un corpus de textes. Par exemple, elle peut aider à comprendre combien de fois deux personnalités politiques sont mentionnées ensemble dans les articles, ou combien de fois une entreprise et un lieu sont cités conjointement

````python
import json

# Fonction pour compter les co-occurrences de deux entités
def compter_cooccurrences(base_donnees, entite1, type1, entite2, type2):
    """
    Compte le nombre de co-occurrences des deux entités dans la même phrase ou document.

    :param base_donnees: Dictionnaire contenant les articles enrichis avec leurs entités.
    :param entite1: Texte de la première entité.
    :param type1: Type de la première entité (ex : 'PER', 'LOC').
    :param entite2: Texte de la deuxième entité.
    :param type2: Type de la deuxième entité (ex : 'LOC', 'ORG').
    :return: Nombre de documents où les deux entités co-occurrent.
    """
    cooccurrence_count = 0

    # Parcourir chaque article dans la base
    for lien, article in base_donnees.items():
        entites = article.get("entities", [])

        # Filtrer les entités correspondantes à entite1 et entite2
        presence_entite1 = any(ent["entite"] == entite1 and ent["type"] == type1 for ent in entites)
        presence_entite2 = any(ent["entite"] == entite2 and ent["type"] == type2 for ent in entites)

        # Si les deux entités sont présentes, on incrémente le compteur
        if presence_entite1 and presence_entite2:
            cooccurrence_count += 1

    return cooccurrence_count

````

# Question 11: Co-occurences d'entités au sein du même document (les plus fréquentes)

Cette fonction peut être utilisée pour analyser les relations entre différentes entités dans un corpus de textes. Par exemple, elle peut aider à comprendre quelles personnalités politiques et lieux sont souvent mentionnés ensemble, ou quelles organisations et lieux sont fréquemment cités conjointement.

````python

from itertools import combinations
from itertools import product
# Fonction pour trouver les paires d'entités de types différents les plus fréquentes
def paires_top_entites(base_donnees, top_per, top_loc, top_org):
    """
    Trouve les paires d'entités (types différents) parmi les top 20 entités par type.
    """
    compteur_paires = Counter()

    # Extraire les noms des entités à partir des top listes
    top_personnes = [entite for entite, _ in top_per]
    top_lieux = [entite for entite, _ in top_loc]
    top_organisations = [entite for entite, _ in top_org]

    # Créer des combinaisons possibles (PER-LOC, PER-ORG, LOC-ORG)
    paires_a_tester = list(product(top_personnes, ["PER"], top_lieux, ["LOC"])) + \
                      list(product(top_personnes, ["PER"], top_organisations, ["ORG"])) + \
                      list(product(top_lieux, ["LOC"], top_organisations, ["ORG"]))

    # Compter les co-occurrences pour chaque paire
    for entite1, type1, entite2, type2 in paires_a_tester:
        cooccurrences = compter_cooccurrences(base_donnees, entite1, type1, entite2, type2)
        if cooccurrences > 0:
            paire = tuple(sorted([(entite1, type1), (entite2, type2)]))
            compteur_paires[paire] = cooccurrences

    return compteur_paires.most_common(10)

````

# Questions 12: Co-occurence d'entités au sein de la même phrase (les plus fréquents)

````python
from collections import Counter
from itertools import product

def compter_cooccurrences_phrase(base_donnees, entite1, type1, entite2, type2):
    """
    Compte le nombre de phrases contenant les deux entités dans un article.
    :param base_donnees: Dictionnaire contenant les articles enrichis avec phrases et entités.
    :param entite1: Texte de la première entité.
    :param type1: Type de la première entité (ex : 'PER', 'LOC').
    :param entite2: Texte de la deuxième entité.
    :param type2: Type de la deuxième entité (ex : 'LOC', 'ORG').
    :return: Nombre de phrases où les deux entités co-occurrent.
    """
    cooccurrence_count = 0

    # Parcourir chaque article
    for article in base_donnees.values():
        phrases = article.get("phrases", [])

        # Parcourir chaque phrase et vérifier la présence des entités
        for phrase in phrases:
            ents_in_phrase = phrase.get("entites", [])
            presence_entite1 = any(ent["entite"] == entite1 and ent["type"] == type1 for ent in ents_in_phrase)
            presence_entite2 = any(ent["entite"] == entite2 and ent["type"] == type2 for ent in ents_in_phrase)

            # Compter si les deux entités sont dans la même phrase
            if presence_entite1 and presence_entite2:
                cooccurrence_count += 1

    return cooccurrence_count

def paires_top_entites_par_phrase(base_donnees, top_per, top_loc, top_org):
    """
    Trouve les paires d'entités co-occurrentes dans les mêmes phrases.
    """
    compteur_paires = Counter()

    # Extraire les noms des entités des top listes
    top_personnes = [entite for entite, _ in top_per]
    top_lieux = [entite for entite, _ in top_loc]
    top_organisations = [entite for entite, _ in top_org]

    # Créer des combinaisons possibles (PER-LOC, PER-ORG, LOC-ORG)
    paires_a_tester = list(product(top_personnes, ["PER"], top_lieux, ["LOC"])) + \
                      list(product(top_personnes, ["PER"], top_organisations, ["ORG"])) + \
                      list(product(top_lieux, ["LOC"], top_organisations, ["ORG"]))

    # Compter les co-occurrences pour chaque paire dans les phrases
    for entite1, type1, entite2, type2 in paires_a_tester:
        cooccurrences = compter_cooccurrences_phrase(base_donnees, entite1, type1, entite2, type2)
        if cooccurrences > 0:
            paire = tuple(sorted([(entite1, type1), (entite2, type2)]))
            compteur_paires[paire] = cooccurrences

    return compteur_paires.most_common(10)

def afficher_top_paires_par_phrase(top_paires):
    """
    Affiche proprement les paires d'entités co-occurrentes dans les mêmes phrases.
    """
    print("\n" + "=" * 60)
    print(f"{'Top 10 des paires d entités dans une même phrase':^60}")
    print("=" * 60)

    print(f"{'N°':<4} {'Entité 1':<25} {'Type 1':<8} | {'Entité 2':<25} {'Type 2':<8} | {'Occurrences':<10}")
    print("-" * 60)

    # Parcourir et afficher chaque paire avec son numéro
    for idx, (paire, freq) in enumerate(top_paires, start=1):
        entite1, type1 = paire[0]
        entite2, type2 = paire[1]
        print(f"{idx:<4} {entite1:<25} {type1:<8} | {entite2:<25} {type2:<8} | {freq:<10}")

    print("=" * 60)
````

# Question 13: Caractéristion d'une paire d'entités au sein d'une même phrase par des verbes

Nous utiliserons la base d'articles enrichie à la question 8. Cette base contient, pour chaque article, les entités nommées des phrases comportant au moins un verbe, ainsi que la liste des verbes associés. Cela permet d'éviter les extractions répétées d'entités et de verbes, rendant l'analyse plus rapide et intuitive.

Cependant, plusieurs verbes peuvent apparaître entre deux entités dans une même phrase, ce qui nécessite une analyse plus poussée pour identifier les verbes qui traduisent réellement la relation entre elles. Cette analyse peut être réalisée en ajustant partiellement notre code pour recupérer l'arbre de dépendance syntaxique.

````python
def extraire_verbes_entre_entites(base_articles, entite1, type1, entite2, type2):
    """
    Trouve les verbes situés entre deux entités spécifiées dans chaque phrase.
    :param base_articles: Dictionnaire contenant les articles enrichis.
    :param entite1: Texte de la première entité.
    :param type1: Type de la première entité (PER, LOC, ORG, etc.).
    :param entite2: Texte de la deuxième entité.
    :param type2: Type de la deuxième entité.
    :return: Liste des verbes lemmatisés entre les deux entités.
    """
    relations = []

    # Parcourir chaque article
    for lien, article in base_articles.items():
        phrases = article.get("phrases", [])

        # Parcourir chaque phrase et vérifier la présence des entités
        for phrase in phrases:
          ents_in_phrase = phrase.get("entites", [])
          verbs_in_phrase = phrase.get("verbes",[])

          # Identifier les deux entités spécifiées
          e1 = next((ent for ent in ents_in_phrase if ent["entite"] == entite1 and ent["type"] == type1),None)
          e2 = next((ent for ent in ents_in_phrase if ent["entite"] == entite2 and ent["type"] == type2),None)

          if e1 and e2:
            # Définir les positions des deux entités
            if e1["start"] < e2["start"]:
              start, end = e1["end"], e2["start"]
            else:
              start, end = e2["end"], e1["start"]
            #start, end = sorted([e1["start"], e2["start"]])

            # Extraire les verbes entre les deux entités
            verbes = [verbe["verbe"] for verbe in verbs_in_phrase if verbe["start"] > start and verbe["end"] < end]
            if verbes:
              relations.append({
                  "phrase": phrase.get("texte"),
                  "verbes": verbes,
                  "article": lien
                        })
    return relations
````
