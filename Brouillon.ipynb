```python
def tableOfContent(text):
    chapter_num = 0
    toc = []
    for line in text:
        if line[:2] == "# ":
            chapter_num += 1
            toc.append(f"{chapter_num}. " + line[1:])
            section_num = 0
        
        if line[:3] == "## ":
            section_num += 1
            toc.append(f"{chapter_num}.{section_num}. " + line[2:])
    return toc
```

```python
text = ["# Monday Dog", "## Darty", "## god", "# Datue", "## Quarto"]
tableOfContent(text)
```

```python
def sameSubstring(s="badc" , t="cadc", K=3) :
    alphabet = {"a":1, "b":2,"c":3,"d":4, "e":5, "f":6, "g":7, "h":8, "i":9,
     "j":10, "k":11, "l":12, "m":13, "n":14, "o":15, "p":16, "q":17, "r":18,
      "s":19, "t":20, "u":21, "v":22, "w":23, "x":24, "y":25, "z":26} 
    max_len = 0
    init_pos = 0
    max_cost = K
    n = len(s)
    for i in range(n):
        cost = abs(alphabet[s[i]] - alphabet[t[i]])
        if cost > max_cost :
            init_pos = i + 1
        else:
            max_cost -= cost
            max_len = max(max_len, i - init_pos + 1)
    return max_len
```

```python
sameSubstring("uaccd", "gbbeg", 4)
```

```python
import numpy as np
def getAnagram(s):
    n = int(len(s) / 2)
    left = np.repeat(0, 10)
    right = np.repeat(0, 10)
    change = 0
    for i in range(n):
        left[int(s[i])] += 1
        right[int(s[i + n])] += 1
    for i in range(10):
        if left[i]>right[i]:
            change += left[i] - right[i]
    return change
```

```python
getAnagram("122122")
```

```python
def maxGamePoints(game_val, k):
    n = len(game_val)
    dp = [0] * n   # dp[i] = max points starting from index i

    # Fill dp from right to left
    for i in range(n - 1, -1, -1):
        next_i = i + k
        if next_i < n:
            dp[i] = game_val[i] + dp[next_i]
        else:
            dp[i] = game_val[i]

    # Answer is the best possible starting position
    return max(dp)
```

```python
game_val = [2, -3, 4, 6, 1]
k = 1
maxGamePoints(game_val, k)
```
